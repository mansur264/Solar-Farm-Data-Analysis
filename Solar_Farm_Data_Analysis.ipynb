```python
import pandas as pd

df = pd.read_csv("Q01_solar_farm.csv")

df.head()
```

```python
# Check missing values
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Check dataset information
df.info()
```

```python
df[[
    "panel_output_kwh",
    "ambient_temperature_c",
    "sunlight_duration_hours"
]].describe()
```

```python
# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Create efficiency ratio
df["efficiency_ratio"] = (
    df["panel_output_kwh"] / df["sunlight_duration_hours"]
)

# Create hour and classify time of day
df["hour"] = df["timestamp"].dt.hour

df["time_of_day"] = pd.cut(
    df["hour"],
    bins=[0, 12, 17, 24],
    labels=["Morning", "Afternoon", "Evening"],
    include_lowest=True
)

df.head()
```

```python
df.head()
```

```python
df.to_csv("cleaned_solar_farm_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
